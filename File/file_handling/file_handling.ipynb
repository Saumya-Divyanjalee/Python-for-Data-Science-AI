```python
# A tution class maintains student attendance recodes using a text file named "attendance.txt".
# Each line of the file contains a student's name and the number of  attendance sessions separated by a comma.
# write a python program to 
# 1.add atleast 10 student recodes to the file
# 2. read all recodes from the file
# 3. display the names of students who's attendence presentage is below 80%
# 4. calculate and display the average attendance percentage of the class.
```

```python
TOTAL_SESSIONS = 20

students = [
    ("Kamal", 17),
    ("Nimal", 14),
    ("Sunil", 18),
    ("Amara", 12),
    ("Saman", 16),
    ("Kumari", 19),
    ("Ravi", 15),
    ("Mali", 13),
    ("Nadeesha", 18),
    ("Kasun", 11)
]

with open("attendance.txt", "w") as file:
    for name, attended in students:
        file.write(f"{name},{attended}\n")

print("All Student Records")
print("-------------------")

with open("attendance.txt", "r") as file:
    for line in file:
        print(line.strip())

print("\nStudents Below 80% Attendance")
print("-----------------------------")

with open("attendance.txt", "r") as file:
    for line in file:
        name, attended = line.strip().split(",")

        percentage = (int(attended) / TOTAL_SESSIONS) * 100

        if percentage < 80:
            print(name, "-", percentage, "%")

total_percentage = 0
count = 0

with open("attendance.txt", "r") as file:
    for line in file:
        name, attended = line.strip().split(",")

        percentage = (int(attended) / TOTAL_SESSIONS) * 100

        total_percentage += percentage
        count += 1

average = total_percentage / count

print("\nAverage Attendance Percentage")
print("-----------------------------")
print(average, "%")
```

```python
# a school stores student examination records in a csv file named "students.csv" 
# the file contains the following columns student_id , name , maths , science ,english
# ex: S001 , Kamal,78,65,80
#     S002 , Nimal,45,55,60
#     S003 , Amal,90,88,95
# write a python program using csv module,
#   1.Read all student records from the csv file 
#   2.calculate total marks and average mask of each student 
#   3.determined each student has pass or fail .pass mark for each subject is 50
#   4.display students who's average marks are above 75
#   5.save fail student details into another csv file named "failed_student.csv"

import csv

# Example student data
students = [
    ["student_id", "name", "maths", "science", "english"],
    ["S001", "Kamal", 78, 65, 80],
    ["S002", "Nimal", 45, 55, 60],
    ["S003", "Amal", 90, 88, 95]
]

# Create students.csv file
with open("students.csv", "w", newline='') as file:
    writer = csv.writer(file)
    writer.writerows(students)

print("students.csv file created")


# Read student records
with open("students.csv", "r", newline='') as file:

    reader = csv.DictReader(file)

    failed_students = []

    print("\nStudent Details")
    print("-------------------------")

    for row in reader:

        student_id = row["student_id"]
        name = row["name"]

        maths = int(row["maths"])
        science = int(row["science"])
        english = int(row["english"])

        # Calculate total and average
        total_marks = maths + science + english
        average_marks = total_marks / 3

        # Determine pass or fail
        if maths >= 50 and science >= 50 and english >= 50:
            result = "Pass"
        else:
            result = "Fail"

            failed_students.append({
                "student_id": student_id,
                "name": name,
                "maths": maths,
                "science": science,
                "english": english
            })

        # Display student details
        print("Student ID :", student_id)
        print("Name       :", name)
        print("Total      :", total_marks)
        print("Average    :", average_marks)
        print("Result     :", result)
        print()

        # Students with average above 75
        if average_marks > 75:
            print(name, "has average above 75")
            print()


# Save failed students to another CSV file
with open("failed_student.csv", "w", newline='') as file:

    fieldnames = ["student_id", "name", "maths", "science", "english"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()

    writer.writerows(failed_students)

print("Failed student details saved to failed_student.csv")
```
