```python
school_clubs = {
    "Robotics": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Chess": ["Charlie", "Eve", "Frank", "Grace"]
}

robo = school_clubs["Robotics"]
chess = school_clubs["Chess"]

print(robo)
print(chess)
print("===========")

ans_1 = set(robo) & set(chess) # intersection

print(ans_1)
print("===========")

robo_set = set(school_clubs["Robotics"])
chess_set = set(school_clubs["Chess"])

print(robo_set)
print(chess_set)

ans_2 = robo_set - chess_set
print(ans_2)

print("================")

ans_3 = robo_set | chess_set
print(ans_3)


                         
```
