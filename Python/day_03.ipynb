Structural Pattern Matching

```python
x = 3
match x:
    case 1:
        print("x is 1")
    case 2:
        print("x is 2")
    case 3:
        print("x is 3")
    case _:
        print("x is something else")
                
                
```

```python
my_list = [2, 4, 3]

match my_list:
    case [x, y]:
        print("x is", x)
        print("y is", y)

    case [x, y, z]:
        print("x is", x)
        print("y is", y)
        print("z is", z)

    case _:
        print("List has a different length")
```

```python
C = 3.14

shape = ["circle", 4]   

match shape:
    case ["circle", r]:
        print("Circle area:", C * r * r)

    case ["square", s]:
        print("Square area:", s * s)

    case ["rectangle", l, w]:
        print("Rectangle area:", l * w)

    case _:
        print("Invalid shape")
```
