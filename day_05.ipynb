```python
def transform_sequence(*numbers, mode="linear"):
    transformed = []
    if mode == "linear":
        transformed = [2 * x + 3 for x in numbers]
    elif mode == "quadratic":
        transformed = [x ** 2 - x for x in numbers]
    elif mode == "factorial_sum":
        transformed = []
        for num in numbers:
            factorial_sum = 0
            factorial = 1
            for i in range(1, num + 1):
                factorial *= i
                factorial_sum += factorial
            transformed.append(factorial_sum)
    
    total = sum(transformed)
    max_value = max(transformed)
    min_value = min(transformed)

    return {
        "transformed": transformed,
        "total": total,
        "max": max_value if transformed else None,
        "min": min_value if transformed else None
    }

print(transform_sequence(1, 2, 3,4))
print(transform_sequence(1, 2, 3, mode="quadratic"))
print(transform_sequence(3, 4, 5, mode="factorial_sum"))
```

```python
def my_fun(**kwargs):
    print(kwargs)
    print(type(kwargs))

my_fun(name="saumya", age=24, city="panadura")
```

```python
def my_fun(**kwargs):
    for key,value in kwargs.items():
        print(key, ":" , value)
        print(type(kwargs))
        
my_fun(name="saumya", age=24, city="panadura")
```

```python
 def calculate_bill(customer, **kwargs):
    subtotal = sum(kwargs.values())

    if subtotal >= 100:
        discount = subtotal * 0.20
    elif subtotal >= 50:
        discount = subtotal * 0.10
    else:
        discount = 0

    after_discount = subtotal - discount
    tax = after_discount * 0.05
    total = after_discount + tax

    return {
        "customer": customer,
        "Subtotal": round(subtotal, 2),
        "Discount": round(discount, 2),
        "Tax": round(tax, 2),
        "Total": round(total, 2)
    }

 
calculate_bill(
        customer="Saumya",
        burger=20,
        pizza=30,
        cola=5
    )
```

```python
users = [
    {"name": "Alice", "age":30},
    {"name": "Bob", "age":20},
    {"name": "charlie", "age":24},
    
]

sorted_users = sorted(users, key=lambda user:user["age"])
print(sorted_users)
```

```python
users = [
    {"name": "Alice", "age": 30},
    {"name": "Bob", "age": 20},
    {"name": "charlie", "age": 24},
]

oldest_user = max(users, key=lambda x: x["age"])

print(oldest_user)
```
