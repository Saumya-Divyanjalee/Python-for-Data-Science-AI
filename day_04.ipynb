```python
 num = int(input("Enter a number: "))

if num == 0:
    print("Factorial: 1")
else:
    fact = 1
    for i in range(1, num + 1):
        fact *= i
    print("Factorial:", fact)
```

```python
num = int(input("Enter a whole number: "))

if num >= 0:
    factorial = 1

    for i in range(1, num + 1):
        factorial *= i

    print("Factorial:", factorial)

else:
    print("Factorial is not defined for negative numbers")
```

```python
my_list = {
    "name": "saumya",
    "city": "colombo",
    "age": 24
}

# Print all keys
for key in my_list.keys():
    print("Key:", key)

print("-----")

# Print all values
for value in my_list.values():
    print("Value:", value)

print("-----")

# Print key + value pairs
for key, value in my_list.items():
    print("Key:", key, "Value:", value)
```

```python
inventory = {
    "Microchips": [50, 12.50],
    "Resistors": [500, 0.10],
    "Capacitors": [200, 0.45]
}

total_store_value = 0

for item in inventory:
    quantity = inventory[item][0]
    price = inventory[item][1]

    total_store_value += quantity * price

print("Total Store Value:", total_store_value)
```

```python
fixed_cost = 500
cost_per_unit = 2
sell_price = 7

for x in range(1, 1001):

    revenue = sell_price * x
    cost = fixed_cost + cost_per_unit * x
    profit = revenue - cost

    print("Units:", x, "Profit:", profit)

    if profit >= 0:
        print("Break-even / Profit starts at:", x)
        break

else:
    print("Always loss (no break-even point)")
```

```python
num1 = int(input("Enter number: "))
num2 = int(input("Enter number: "))

for i in range(2, min(num1, num2) + 1):
    if num1 % i == 0 and num2 % i == 0:
        print("Not Coprime")
        break
else:
    print("Coprime")
```

```python
num1 = int(input("Enter number: "))
num2 = int(input("Enter number: "))

if num1 < num2:
    minimum = num1
else:
    minimum = num2

for i in range(2, minimum + 1):
    if num1 % i == 0 and num2 % i == 0:
        print("Not Coprime")
        break
else:
    print("Coprime")
```

```python
num = int(input("Enter an integer: "))

reverse = 0

while num > 0:
    temp = num % 10
    reverse = reverse * 10 + temp
    num = num // 10

print(reverse)
```

```python
x1, y1 = 1, 2
x2, y2 = 4, 6

distance = ((x2 - x1)**2 + (y2 - y1)**2) ** 0.5

mid_x = (x1 + x2) / 2
mid_y = (y1 + y2) / 2

 
print("Distance:", distance)
print("Midpoint:", (mid_x, mid_y))
```

```python
def analyze(x1, y1, x2, y2):

    
    distance = ((x2 - x1)**2 + (y2 - y1)**2) ** 0.5

     
    mid_x = (x1 + x2) / 2
    mid_y = (y1 + y2) / 2

     
    return distance, mid_x, mid_y

d, mx, my = analyze(1, 2, 4, 6)

print("Distance:", d)
print("Midpoint:", (mx, my))
```

```python
def power(base, exponent=2):
    return base ** exponent

 
print("5 power 2 =", power(5))
print("5 power 3 =", power(5, 3))
print("2 power 4 =", power(2, 4))
```
