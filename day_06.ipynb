```python
class Circle:
    PI = 3.14
    
    def __init__(self, r):
        self.r = r

    
    def calculate_circumference(self):
        circumference = 2 * Circle.PI * self.r
        print("Circumference =", circumference)

   
    def calculate_area(self):
        area = Circle.PI * self.r * self.r
        print("Area =", area)


 
circle1 = Circle(5)

 
circle1.calculate_circumference()
circle1.calculate_area()
```

```python
class Vehicle:
    has_wheels = True
    def __init__(self):
        print("Super class constructor is called")
        
    def display_info(self):
        print("This is a super class method")
        
        # not counstructor in sub class, so it called super class constructor that reason it called super class constructor and print it, but sub class has constructor it print
        # now sub class has constructor now it auto called and print it
class Car(Vehicle):
    def __init__(self):
        print("Sub class constructor is called")
        
    def display_info(self):
        print("This is a sub class method")
       
car_1 = Car()
car_1.display_info() # method overrideing
    
```

```python
 
class Employee:
    def __init__(self, name, base_salary):
        self.name = name
        self.base_salary = base_salary
        print("super class constructor is called")

    def calculate_pay(self):
        return self.base_salary
    
class Manager(Employee):
    def __init__(self, name, base_salary, bonus):
        super().__init__(name, base_salary)
        self.bonus = bonus
        print("sub class constructor is called")

    def calculate_pay(self):
        return self.base_salary + self.bonus


 
e1 = Employee("Saumya", 50000)
print(e1.calculate_pay())

m1 = Manager("Divyanjalee", 70000, 20000)
print(m1.calculate_pay())
```

```python
# protected - developer can understand this is protected, this is best practise but not recommanded , can't access out side
# recommand to access -> subclass access . It is recommanded way
 
class Employee:
    def __init__(self, name):
        self._name = name # protected attribute -> _name (use one underscore , it mean that variable is protected)
        
       
    # protected method 
    def _display_info(self):
        pass
        
        
employee_1 = Employee("saumya")
print(employee_1._name)
```

```python
# private 
 
class Employee:
    def __init__(self, name):
        self.__name = name # private attribute use __ two underscores
        
    def __display_info(self):
        pass
        
employee_1 = Employee("saumya")
print(employee_1.__name) # can't access, because __name is private attribute so can't access class outside, only access can class inside
```

```python
# private

class Employee:
    def __init__(self, name):
        self.__name = name   # private attribute

    # getter method
    def get_name(self):
        return self.__name

    # setter method
    def set_name(self, new_name):
        self.__name = new_name


employee_1 = Employee("saumya")
print("Before:", employee_1.get_name())
employee_1.set_name("divyanjalee")
print("After:", employee_1.get_name())
```

```python
class Fraction:
    def __init__(self, numerator, denominator):
        self.__numerator = numerator

        if denominator != 0:
            self.__denominator = denominator
             
        else:
            print("denominator cannot be zero")
            self.__denominator = 1
           
    def set_denominator(self, new_denominator): # update denominator
        if new_denominator != 0:
            self.__denominator = new_denominator
        else:
            print("denominator cannot be zero")

    def get_decimal_value(self):
        return self.__numerator / self.__denominator


 
fraction_1 = Fraction(10, 0)
print("Decimal Value:", fraction_1.get_decimal_value())
fraction_1.set_denominator(5)
print("New Decimal Value:", fraction_1.get_decimal_value())
```

```python
class Shape:
    def __init__(self, **kwargs):
         for key, value in kwargs.items():
            setattr(self, key, value)

class Cylinder(Shape):
    def __init__(self, radius, height, **kwargs):
        self.radius = radius
        self.height = height

        super().__init__(**kwargs)

    def calculate_volume(self):
        pi = 3.14
        return pi * (self.radius ** 2) * self.height


 
cylinder_1 = Cylinder(5, 10, unit="cm", material="metal")
print("Volume:", cylinder_1.calculate_volume())

print("Unit:", cylinder_1.unit)
print("Material:", cylinder_1.material)
```
